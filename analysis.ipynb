# Регіональна економічна нерівність в Україні, 2021

```
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/regional_economic_data_2021.csv')
df.head()
```

```
df.describe(include='all')
```

```
df.sort_values('avg_monthly_wage_uah', ascending=False)[['region','avg_monthly_wage_uah']].head(10)
```

```
print('Кореляція зарплати та урбанізації:', round(df['avg_monthly_wage_uah'].corr(df['urban_share_pct']), 2))
print('Кореляція зарплати та населення:', round(df['avg_monthly_wage_uah'].corr(df['population']), 2))
```

```
plot_df = df.sort_values('avg_monthly_wage_uah')
plt.figure(figsize=(10,8))
plt.barh(plot_df['region'], plot_df['avg_monthly_wage_uah'])
plt.xlabel('Середньомісячна зарплата, грн')
plt.title('Зарплата за регіонами, 2021')
plt.tight_layout()
plt.show()
```

```
plt.figure(figsize=(8,6))
plt.scatter(df['urban_share_pct'], df['avg_monthly_wage_uah'])
plt.xlabel('Частка міського населення, %')
plt.ylabel('Середньомісячна зарплата, грн')
plt.title('Зарплата та урбанізація')
plt.show()
```
